## Aim
To reduce the 4-dimensional **Iris dataset** down to **2 Principal Components** using **Principal Component Analysis (PCA)** and visualize the classes.

## Concept
PCA transforms correlated features into a smaller set of uncorrelated **Principal Components (PCs)** that preserve as much variance (information) as possible.


```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Step 2: Load Iris dataset and standardize features
iris = load_iris()
X = iris.data
y = iris.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Original shape:", X.shape)
```

```python
# Step 3: Apply PCA to reduce from 4 features to 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Reduced shape:", X_pca.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_.round(4))
print("Total variance kept:", round(sum(pca.explained_variance_ratio_) * 100, 2), "%")
```

```python
# Step 4: Plot the 2D PCA projection
plt.figure(figsize=(7, 5))
for class_id, name in enumerate(iris.target_names):
    plt.scatter(X_pca[y == class_id, 0], X_pca[y == class_id, 1], label=name, s=50)

plt.title("Iris Dataset after PCA (4D -> 2D)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Conclusion
- PCA reduced the Iris dataset from **4 features to 2 Principal Components** while keeping over **95%** of the original variance.
- The 2D scatter plot clearly separates *Setosa* from *Versicolor* and *Virginica*.
